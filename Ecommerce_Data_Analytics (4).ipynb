```python
import pandas as pd
import glob

file = glob.glob("ecommerce_data/*.csv")[0]
df = pd.read_csv(file)

df.head()
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
df['order_date'] = pd.to_datetime(df['order_date'])

print(df.dtypes)
```

```python
print("Total Orders:", len(df))
print("Total Revenue:", df['revenue'].sum())
print("Total Quantity Sold:", df['quantity'].sum())
print("Average Revenue:", df['revenue'].mean())
```

```python
category_sales = df.groupby('product_category')['revenue'].sum()

print(category_sales)
```

```python
region_sales = df.groupby('region')['revenue'].sum()

print(region_sales)
```

```python
payment_counts = df['payment_method'].value_counts()

print(payment_counts)
```

```python
print("Most Used Payment Method:")
print(payment_counts.idxmax())

print("Number of Orders:")
print(payment_counts.max())
```

```python
average_rating = df['customer_rating'].mean()

print("Average Customer Rating:", average_rating)
```

```python
rating_counts = df['customer_rating'].value_counts().sort_index()

print(rating_counts)
```

```python
average_delivery = df['delivery_days'].mean()

print("Average Delivery Days:", average_delivery)
```

```python
discount_analysis = df.groupby('discount')['revenue'].sum()

print(discount_analysis)
```

```python
monthly_sales = df.groupby(
    df['order_date'].dt.to_period('M')
)['revenue'].sum()

print(monthly_sales)
```

```python
best_month = monthly_sales.idxmax()
highest_month_sales = monthly_sales.max()

print("Best Sales Month:", best_month)
print("Highest Monthly Revenue:", highest_month_sales)
```

```python
region_sales = df.groupby('region')['revenue'].sum().sort_values(ascending=False)

print("Region-wise Revenue:")
print(region_sales)
```

```python
payment_sales = df.groupby('payment_method')['revenue'].sum().sort_values(ascending=False)

print("Payment Method-wise Revenue:")
print(payment_sales)
```

```python
category_quantity = df.groupby('product_category')['quantity'].sum().sort_values(ascending=False)

print("Category-wise Quantity Sold:")
print(category_quantity)
```

```python
customer_orders = df['customer_id'].value_counts()

print("Top 10 Customers by Number of Orders:")
print(customer_orders.head(10))
```

```python
customer_revenue = df.groupby('customer_id')['revenue'].sum().sort_values(ascending=False)

print("Top 10 Customers by Revenue:")
print(customer_revenue.head(10))
```
